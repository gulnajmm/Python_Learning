```python
#1. Single inheritance
class Bank:
    def interest(self):
        print("Bank interest: 5%")

class SBI(Bank):
    def interest(self):          # overrides Bank.interest
        print("SBI interest: 7%")

SBI().interest()   # SBI interest: 7%
```

```python
#2. Multilevel inheritance

#Each level can override, and the lowest class that defines the method wins.
class GrandParent:
    def tax(self):
        print("GrandParent tax: 10%")

class Parent(GrandParent):
    def tax(self):               # overrides GrandParent.tax
        print("Parent tax: 5%")

class Child(Parent):
    def tax(self):               # overrides Parent.tax
        print("Child tax: 2%")

Child().tax()    # Child tax: 2%
Parent().tax()   # Parent tax: 5%
GrandParent().tax()   # GrandParent tax: 10%
```

```python
#Multilevel overriding with super() at every level:
class A:
    def show(self):
        print("A")

class B(A):
    def show(self):
        super().show()
        print("B")

class C(B):
    def show(self):
        super().show()
        print("C")

C().show()
# A
# B
# C
```

```python
#3. Multiple inheritance
#When both parents have the same method, the left-most parent wins.
class Father:
    def car(self):
        print("Father's car: BMW")

class Mother:
    def car(self):
        print("Mother's car: Audi")

class Child(Mother,Father):
    pass

#Child().car()    #left most in () wins
class Child(Father, Mother):
    def car(self):
        print("Child's car: Tesla")

Child().car()    # Child's car: Tesla

```

```python
class Child(Father, Mother):
    def car(self):
        Father.car(self)
        Mother.car(self)
        print("Child's car: Tesla")

Child().car()
# Father's car: BMW
# Mother's car: Audi
# Child's car: Tesla
```

```python
#4. Hierarchical inheritance

#Each child overrides the same parent method in its own way (this is the classic polymorphism example).
class Shape:
    def area(self):
        print("Area of shape")

class Circle(Shape):
    def area(self):              # overrides
        print("Area = π × r × r")

class Square(Shape):
    def area(self):              # overrides
        print("Area = side × side")

for s in (Shape(), Circle(), Square()):
    s.area()
# Area of shape
# Area = π × r × r
# Area = side × side
```

```python
#5. Hybrid inheritance
class A:
    def show(self):
        print("A's show")

class B(A):
    def show(self):              # overrides A
        print("B's show")

class C(A):
    def show(self):              # overrides A
        print("C's show")

class D(B, C):
    pass

D().show()    # B's show  (MRO: D -> B -> C -> A)
```

```python
#With super(), every class runs once, following the MRO:
class A:
    def show(self):
        print("A")

class B(A):
    def show(self):
        print("B")
        super().show()

class C(A):
    def show(self):
        print("C")
        super().show()

class D(B, C):
    def show(self):
        print("D")
        super().show()

D().show()
# D
# B
# C
# A
```

# Summary

# Type	    Which method runs?

# Single	  Child's version

# Multilevel	Nearest class (lowest) that defines it

# Multiple	Left-most parent in the class list

# Hierarchical	Each child runs its own version

# Hybrid	First match in the MRO

```python
#Example 1: Basic
class Animal:
    def eat(self):
        print("Animal eats food")

class Dog(Animal):
    def bark(self):
        print("Dog barks")

d = Dog()
d.eat()    # Animal eats food  (inherited)
d.bark()   # Dog barks         (own method)
```

```python
#Example 2: Inheriting variables
class Person:
    species = "Human"

class Student(Person):
    college = "ABC College"

s = Student()
print(s.species)   # Human         (from Person)
print(s.college)   # ABC College   (own)
```

```python
#Example 3: Method overriding
class Bank:
    def interest(self):
        print("Bank interest is 5%")

class SBI(Bank):
    def interest(self):
        print("SBI interest is 7%")

b = SBI()
b.interest()   # SBI interest is 7%
```

```python
#Example 4: Using super()-->The child runs the parent's method as well as its own.
class Phone:
    def call(self):
        print("Calling...")

class SmartPhone(Phone):
    def call(self):
        super().call()          # run parent's version first
        print("Video call enabled")

p = SmartPhone()
p.call()
# Calling...
# Video call enabled
```

```python
#Example 5: Constructor (__init__) with super()
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

class Manager(Employee):
    def __init__(self, name, salary, team_size):
        super().__init__(name, salary)   # set parent's data
        self.team_size = team_size

    def show(self):
        print(self.name, self.salary, self.team_size)

m = Manager("Ravi", 80000, 5)
m.show()   # Ravi 80000 5

'''
The child gets all methods and variables of the parent.
The child can add new ones or override existing ones.
super() lets the child reuse the parent's code instead of rewriting it.'''
```
