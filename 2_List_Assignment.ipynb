```python
#3.  Write a Python program to move all zeros to the end of a list without changing the order of non-zero elements.
Input =  [0, 1, 0, 3, 12]
Input_2 = []
zero_count = 0
for i in Input:
  if i == 0:
    zero_count+=1
  else:
    Input_2.append(i)

for i in range(zero_count):
    Input_2.append(0)

print(Input_2)
```

```python
#9.  Write a Python program to find common elements between two lists while preserving duplicate occurrences.
#Input  = [1, 2, 2, 3]
#List 2 = [2, 2, 4]
#Common elements = [2, 2]

List1 = [1, 2, 2, 3]
List2 = [2, 2, 4]
common = []
temp = List2.copy()   # then use temp in the loop instead of List2

for i in List1:
    if i in List2:
        common.append(i)
        temp.remove(i)

print(common)   # [2, 2]
```

```python
#10.  Write a Python program to transpose a two-dimensional list.
#Input  [[1, 2, 3], [4, 5, 6]]
# [[1, 4], [2, 5], [3, 6]]

matrix = [[1, 2, 3], [4, 5, 6]]
result = []

for j in range(len(matrix[0])):          # each column position: 0, 1, 2
    new_row = []
    for i in range(len(matrix)):         # each row: 0, 1
        new_row.append(matrix[i][j])
    result.append(new_row)

print(result)   # [[1, 4], [2, 5], [3, 6]]
```
