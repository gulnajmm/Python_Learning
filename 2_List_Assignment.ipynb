```python
#2. Write a Python program to find the sum and average of list elements.
in_list1 = [10, 20, 30, 40]
sum = 0
for i in in_list1:
  sum = sum + i
  avg = sum/len(in_list1)

print("sum is  ",sum)
print("avg is  ",avg)
```

```python
#4.  Write a Python program to count even and odd numbers in a list.
input_list = [1, 2, 3, 4, 5, 6]
even_count = 0
odd_count = 0
for num in input_list:
    if num % 2 == 0:
        even_count +=1
    else:
        odd_count +=1
print("Even numbers count:", even_count)
print("Odd numbers count:", odd_count)
```

```python
#6.  Write a Python program to reverse a list without using the reverse() method.
list1 = [10, 20, 30, 40]
#list1.reverse()
#print(list1)
list1= list1[::-1] #-1 is from right to left index
print(list1)
```

```python
#using loop also
list1 = [10, 20, 30, 40]
for i in range(len(list1) // 2):
    list1[i], list1[-i-1] = list1[-i-1], list1[i]
print(list1)
```

```python
list1 = [10, 20, 30, 40]
left = 0
right = len(list1) - 1

while left < right:
    list1[left], list1[right] = list1[right], list1[left]
    left += 1
    right -= 1

print(list1)
```

```python
#13.  Write a Python program to separate positive and negative numbers from a list.
in_list= [-5, 3, -2, 7, 0, -1]
pos_list=[]
neg_list=[]
for i in in_list:
  if i< 0 :
    neg_list.append(i)
  elif i>0:
    pos_list.append (i)
print("positive list is ",pos_list)
print("negative list is ",neg_list)
```

```python
#14.  Write a Python program to find common elements between two lists.
List_1 = [1, 2, 3, 4]
List_2 = [3, 4, 5, 6]
common_elements = []
for i in List_1:
  if i in List_2:
    common_elements.append(i)
print(common_elements)
```
