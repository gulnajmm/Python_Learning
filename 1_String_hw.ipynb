```python
#7. Write a Python program to check whether two strings are anagrams.
'''
listen
silent
'''
str1 = "listen"
str2 = "silent"

str1 = str1.lower()
str2 = str2.lower()

if sorted(str1) == sorted(str2):
    print("Anagram")
else:
    print("Not Anagram")

```

```python
#15. Write a Python program to remove duplicate characters from a string while preserving order.
str_input = "programming"
new_str = ""

for i in str_input:
    if i not in new_str:
        new_str = new_str + i

print(new_str)
```

```python
#16. Write a Python program to count uppercase letters, lowercase letters, digits, and special characters in a string
str1 = "LihaTech@2026"
upper_count = 0
lower_count = 0
digit_count = 0
special_count = 0

for ch in str1:
    if ch.isupper():
        upper_count = upper_count+1
    elif ch.islower():
        lower_count = lower_count+1
    elif ch.isdigit():
        digit_count= digit_count+1
    elif not ch.isalnum():
      special_count= special_count +1

print("Uppercase =  ",upper_count,"Lowercase = ",lower_count,"Digit_count =  ",digit_count,"special_count =  ",special_count)
```
