```python
#4.  Write a Python program to find all starting positions of a substring, including overlapping occurrences.
#Input String = aaaaa Substring = aa Expected Output Positions = [0, 1, 2, 3]
def find_all_positions(text, sub):
    positions = []
    n, m = len(text), len(sub)
    for i in range(n - m + 1):
        if text[i:i + m] == sub:   # slice a window of length m and compare
            positions.append(i)
    return positions


print(find_all_positions("aaaaa", "aa"))  # [0, 1, 2, 3]
```

```python
#2.  Write a Python program to find the longest palindromic substring in a given string
#Input babad Expected Output One longest palindromic substring = bab
given_str = "babad"

longest = ""

for center in range(len(given_str)):
    left = center
    right = center

    while left >= 0 and right < len(given_str) and given_str[left] == given_str[right]:
        sub = given_str[left:right + 1]
        if len(sub) > len(longest):
            longest = sub
        left -= 1
        right += 1

print("One longest palindromic substring =", longest)
```

```python
#3.  Write a Python program to encrypt a string using a Caesar cipher with a given shift.
#Input Text = Hello World Shift = 3 Expected Output Encrypted text = Khoor Zruog

Input_Text = "Hello World"
Shift = 3

lower = "abcdefghijklmnopqrstuvwxyz"
upper = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"

result = ""

for ch in Input_Text:
    if ch in lower:
        pos = lower.index(ch)
        result += lower[(pos + Shift) % 26]
    elif ch in upper:
        pos = upper.index(ch)
        result += upper[(pos + Shift) % 26]
    else:
        result += ch      # space or symbol stays the same

print("Encrypted text =", result)
```

```python
#4.  Write a Python program to decode a run-length encoded string. Input a3b2c4 Expected Output Decoded string = aaabbcccc

Input = "a12b3"

result = ""
letter = ""
num = ""

for ch in Input:
    if ch.isalpha():
        result += letter * int(num or 0)
        letter = ch
        num = ""
    else:
        num += ch

result += letter * int(num)

print("Decoded string =", result)   # aaaaaaaaaaaabbb
```

```python
#5.  Write a Python program to check whether brackets in an expression are balanced. Input {[()()]} Expected Output Balanced = True
Input = "{[()()]}"

pairs = {')': '(', ']': '[', '}': '{'}
stack = []
balanced = True

for ch in Input:
    if ch in "([{":
        stack.append(ch)
    elif ch in ")]}":
        if len(stack) > 0 and stack[-1] == pairs[ch]:
            stack.pop()
        else:
            balanced = False
            break

if len(stack) > 0:
    balanced = False

print("Balanced =", balanced)
```

```python
#6.  Write a Python program to find the longest common prefix among a list of strings. Input ['flower', 'flow', 'flight'] Expected Output Longest common prefix = fl
words = ['flower', 'flow', 'flight']

prefix = words[0]

for word in words:
    while not word.startswith(prefix):
        prefix = prefix[:-1]     # remove the last letter

print("Longest common prefix =", prefix)
```

```python
#7.  Write a Python program to group a list of words into anagram groups. Input ['eat', 'tea', 'tan', 'ate', 'nat', 'bat'] Expected Output [['eat', 'tea', 'ate'], ['tan', 'nat'], ['bat']] (group order may vary)

words = ['eat', 'tea', 'tan', 'ate', 'nat', 'bat']

groups = {}

for word in words:
    key = "".join(sorted(word))
    if key not in groups:
        groups[key] = []
    groups[key].append(word)

print(list(groups.values()))
```

```python
#9.  Write a Python program to determine whether a string is made by repeating a smaller substring. Input Expected Output abcabcabcabc Repeated pattern = abc Repeat count = 4
Input = "abcabcabcabc"

n = len(Input)
pattern = ""
count = 1

for k in range(1, n // 2 + 1):
    if n % k == 0:                     # k must divide the length evenly
        piece = Input[:k]              # first k characters
        if piece * (n // k) == Input:  # repeat it and compare
            pattern = piece
            count = n // k
            break                      # smallest piece found, stop

if pattern != "":
    print("Repeated pattern =", pattern)
    print("Repeat count =", count)
else:
    print("No repeated pattern")
```

```python
#10.  Write a Python program to find the smallest substring that contains all characters of a given pattern. Input Expected Output String = ADOBECODEBANC Pattern = ABC Minimum window = BANC
String = "ADOBECODEBANC"
Pattern = "ABC"

best = ""

for i in range(len(String)):
    for j in range(i + 1, len(String) + 1):
        sub = String[i:j]

        ok = True
        for ch in Pattern:
            if sub.count(ch) < Pattern.count(ch):
                ok = False

        if ok and (best == "" or len(sub) < len(best)):
            best = sub

print("Minimum window =", best)
```
